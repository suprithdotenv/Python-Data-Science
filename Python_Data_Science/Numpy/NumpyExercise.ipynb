### Valid Sudoku

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],

    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],

    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
## For a valid Susoku the sum of each and every row is equal to 45
```

```python
## Row axis is 1 and Column axis is 0
```

```python
b = np.sum(s,axis=1)
b
```

```python
b == 45
```

```python
for i in b:
    if i != 45:
        print("Sudoku is not valid")
        break
else:
    print("for rows it is valid")
```

```python
c = np.sum(s,axis=1)
c
```

```python
for i in c:
    if i != 45:
        print("Sudoku is not valid")
        break
else:
    print("for column it is valid")
```

```python
## Valid suduko : 3 * 3 blocks sum is also 45
```

```python
for i in range(0,9,3):
    for j in range(0,9,3):
       print(np.sum(s[i:i+3,j:j+3]))
        
```

```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],  # Student 1
    [19, 92, 88],  # Student 2
    [17, 76, 95],  # Student 3
    [18, 65, 70],  # Student 4
    [20, 90, 85]   # Student 5
])
```

```python
# Get the shape of the Matrix
data.shape
```

```python
#Find the average age of students
s = data[:,0]
s.mean()
```

```python
#Find the highest Science marks
science_marks = data[:,2]
science_marks.max()
```
