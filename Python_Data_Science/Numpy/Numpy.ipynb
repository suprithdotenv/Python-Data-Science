## range() vs np.linspace()
### - range(): Generates integers using a specified step size. Stop value is excluded.
###  range(1, 10, 2)  # 1, 3, 5, 7, 9
### - np.linspace(): Generates a specified number of evenly spaced values. Stop value is included by default.
###  np.linspace(1, 10, 5)  # 1, 3.25, 5.5, 7.75, 10

```python
## Linespace returns evenely spaced numbers in the provided interval
```

```python
arr = np.linspace(1,5,5)
arr
```

```python
arr = np.linspace(1,5,10)
arr
```

## Deep and Shallow copy

```python
arr  = np.arange(1,21)
arr
```

```python
slice = arr[0:5]

#Broadcasting in slice
slice = slice * 10
slice
```

```python
arr
"""**This is shallow copy***"""
#there is no changes in main array even after slicing and broadcasting
```

```python
b = arr
b[0] = 99

arr

#In this case it gets changed because b is just the reference for a
```

```python
np.hsplit(c,4)  
```
